Practical Task: ML Model Versioning & Automated Testing

Build a mini MLOps workflow using Python, MLflow, and pytest.

Requirements:

1. MLflow:Train a Decision Tree classifier, log its accuracy and model parameters, and register the trained model.
2. Testing: Write two pytest test cases to validate model predictions and ensure prediction output is not empty.
3. CI/CD: Create a GitHub Actions workflow that automatically executes pytest whenever code is pushed to the repository.

Expected Output: A versioned ML model, two passing test cases, and an automated GitHub Actions workflow.

git init



```python
# train.py
from sklearn.datasets import iris_dataset
from sklearn.tree import DecisionTreeClassifier
import joblib 
import mlflow
from sklearn.model_selection import train_test_split

X, y = iris_dataset(return_X_y=True)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

n_est=100

with mlflow.start_run():
    model = DecisionTreeClassifier(n_estimator=n_est)
    model.fit(X_train, y_train)
    pred = model.score(X_test, y_test)
    mlflow.log_param("n_estimator", n_est)
    mlflow.log_outputs("predictions", pred)
    mlflow.sklearn.log_model(model)

    joblib.dump(model)
```
