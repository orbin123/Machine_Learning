Duration: 6 minutes

Create a Python program using Pandas and NLTK or TextBlob to perform sentiment analysis on customer reviews.

Dataset:

```
reviews = [    "The product quality is excellent and delivery was fast",    "Very disappointed with the poor customer service",    "The product is okay, nothing special",    "Amazing experience, I will definitely buy again",    "The item arrived damaged and customer support was unhelpful",    "Good value for money but delivery was slow",    "Absolutely terrible quality, waste of money",    "I am happy with the purchase"]
```

Requirements:

1. Convert reviews into lowercase and remove punctuation.
2. Perform sentiment analysis and classify each review as Positive, Negative, or Neutral.
3. Store the results in a Pandas DataFrame with `review`, `sentiment`, and `polarity_score` columns.
4. Calculate the percentage of positive, negative, and neutral reviews.
5. Identify the overall customer sentiment distribution.

````python
reviews = [    "The product quality is excellent and delivery was fast",    "Very disappointed with the poor customer service",    "The product is okay, nothing special",    "Amazing experience, I will definitely buy again",    "The item arrived damaged and customer support was unhelpful",    "Good value for money but delivery was slow",    "Absolutely terrible quality, waste of money",    "I am happy with the purchase"]
````

````python
import re
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer

clean_text = lambda x: map(x, re.findall(r"w\s", "").lower())

for sent in reviews:
    sent = clean_text(sent)

sentmentanalyzer = SentimentIntensityAnalyzer()
scores = [sentmentanalyzer().polarity_scores(text) for text in reviews]

data = {
    sent: score for sent, score in zip(reviews, scores)
}
df = pd.DataFrame(data)
df.head()
````
