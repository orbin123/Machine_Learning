## 7. Attention and Transformer Models

### 1. Build a simple Transformer from scratch.

```python
import tensorflow as tf
from tensorflow.keras import layers

class TransfomerBlock(layers.Layer):
    def __init__(self, d_model, num_heads):
        super().__init__()

        self.attention = layers.MultiHeadAttention(
            num_heads=num_heads,
            key_dim=d_model // num_heads,
        )
        self.feed_forward = tf.keras.Sequential([
            layers.Dense(4 * d_model, activation="relu"),
            layers.Dense(d_model)
        ])

        self.norm1 = layers.LayerNormalization()
        self.norm2 = layers.LayerNormalization()

    def call(self, x):
        # Each token attends to the other tokens
        attended = self.attention(query=x, key=x, value=x)
        x = self.norm1(x + attended)

        # Transform each token independently
        fed = self.feed_forward(x)
        return self.norm2(x + fed)

# Batch=2, sequence_length=5, vector_size=32
x = tf.random.normal((2, 5, 32))

block = TransfomerBlock(d_model=32, num_heads=4)
output = block(x)

print(output.shape) # (2, 5, 32)
```
