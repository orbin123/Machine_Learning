## Question 2 — Valid Parentheses ⭐⭐⭐⭐⭐

This is one of the most common beginner DSA interview questions and tests your understanding of **Stacks**.

### Problem

Given a string containing only:

```text
( ) { } [ ]
```

Determine whether the parentheses are valid.

A string is valid if:

- Every opening bracket has a matching closing bracket.
- Brackets close in the correct order.

### Example 1

```python
Input:
"({[]})"

Output:
True
```

### Example 2

```python
Input:
"([)]"

Output:
False
```

### Requirement

- Solve it using a **Stack** (Python list can be used as a stack).
- Return `True` if the string is valid, otherwise `False`.

````python
def valid_parenthesis(parenthesis_string):

    stack = []

    pairs = {
        ")": "(",
        "]": "[",
        "}": "{"
    }

    for ch in parenthesis_string:
        if ch in "({[":
            stack.append(ch)
        else:
            if not stack or stack[-1] != pairs[ch]:
                return False 

            stack.pop()

    return True 

valid_parenthesis("([)]")
````

## Question 4 — End-to-End Customer Segmentation using K-Means ⭐⭐⭐⭐⭐

You are given a customer dataset such as the **Mall Customers dataset**.

Build an **unsupervised learning pipeline** using K-Means clustering.

Your code should:

- Load the dataset
- Show basic EDA:
  - `info()`
  - missing values
  - descriptive statistics
- Clean the data if needed
- Select numerical features
- Scale them using `StandardScaler`
- Use the **Elbow Method** to choose `k`
- Train `KMeans`
- Add cluster labels to the DataFrame
- Visualize the clusters using a scatter plot
- Briefly interpret the clusters

````python
import pandas as pd

df = pd.read_csv('Mall_Customers.csv')
df.head()
````

````python
df.info()
````

````python
df.isnull().sum()
````

````python
df.describe()
````

````python
df.drop_duplicates(inplace=True)
````

````python
features = ["Gender", "Age", "Annual Income (k$)", "Spending Score (1-100)"]

from sklearn.preprocessing import LabelEncoder, StandardScaler

le = LabelEncoder()

df["Gender"] = le.fit_transform(df["Gender"])

X = df[features]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
````

````python
from sklearn.cluster import KMeans 

wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

import matplotlib.pyplot as plt 

plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("No of Clusters")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
````

````python
kmeans = KMeans(n_clusters=5, random_state=42)

clusters = kmeans.fit_predict(X_scaled)
````

````python
df["Cluster"] = clusters
````
