```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None 

head = Node(10)
head.next = Node(20)
head.next.next = Node(30)

print(head.data)
print(head.next.data)
```

### 2. Construct a doubly linked list

```python
class Node:
    def __init__(self, data):
        self.data = data 
        self.next = None 
        self.next = None

a = Node(10)
b = Node(20)

a.next = b
b.prev = a 

print(a.data, a.next.data)
```

 Q19. Implement linear search

```python
arr = [10, 20, 30, 40, 50]

for i in range(len(arr)):
    if arr[i] == 30:
        print("Found at", i)
        break
```

```python
arr = [10, 20, 30, 40, 50]
target = 40

low = 0
high = len(arr) -1

while low <= high:
    mid = (low+high) // 2

    if arr[mid] == target:
        print("Found at", mid)
        break 
    elif arr[mid] < target:
        low = mid + 1

    else:
        high = mid -1
```

 Q21. Find factorial using recursion

```python
def factorial(n):
    if n == 0:
        return 1

    return n * factorial(n-1)

print(factorial(5))
```

Print numbers using recursion

```python
def print_num(n):
    if n == 0:
        return 

    print(n)
    print_num(n-1)

print_num(5)
```

## Q1. Bubble Sort

**Idea:** Compare adjacent elements and swap if they are in the wrong order.

```python
def bubble(a):
    for i in range(len(a)):
        for j in range(len(a) - i - 1):
            if a[j] > a[j+1]:
                a[j], a[j+1] = a[j+1], a[j]
    return a

a = [5, 2, 4, 1, 3]
print(bubble(a))
```

## Q2. Insertion Sort

**Idea:** Take each element and insert it into its correct position.

```python
def insertion(a):
    for i in range(1, len(a)):
        x = a[i]
        j = i -1 

        while j>= 0 and a[j] > x:
            a[j+1] = a[j]
            j -= 1

        a[j+1] = x 

    return a 

a = [5, 2, 4, 1, 3]
print(insertion(a))
```

## Q3. Selection Sort

**Idea:** Find the smallest element and put it in the current position.

```python
def selection(a):
    for i in range(len(a)):
        m = i 

        for j in range(i+1, len(a)):
            if a[j] < a[m]:
                m = j

        a[i], a[m] = a[m], a[i]

    return a

a = [5, 2, 4, 1, 3]
print(selection(a))
```

```python
class Hash:
    def __init__(self):
        self.t = [[] for _ in range(5)]

    def put(self, k, v):
        self.t[k%5].append((k, v))

    def get(self, k):
        for key, value in self.t[k%5]:
            if key == k:
                return value 

h = Hash()
h.put(12, "A")
h.put(7, "B")

print(h.get(12))
```

## Q5. Stack — PUSH, POP, DISPLAY

**Stack = LIFO**

> Last In, First Out

```python
stack = []

def push(x):
    stack.append(x)

def pop():
    if stack:
        return stack.pop()
    return "Empty"

def display():
    print(stack)

push(10)
push(20)
push(30)

display()
print(pop())
display()
```

# Q6. Queue — ENQUEUE, DEQUEUE, DISPLAY

```python
from collections import deque 

q = deque()

def enqueue(x):
    q.append(x)

def dequeue():
    if q:
        return q.popleft()
    return "Empty"

def display():
    print(list(q))

enqueue(10)
enqueue(20)
enqueue(30)

display()
print(dequeue())
display()
```

# Q7. Binary Search Tree

```python
class Node:
    def __init__(self, x):
        self.x = x 
        self.left = None 
        self.right = None 

def insert(r, x):
    if not r:
        return Node(x)

    if x < r.x:
        r.left = insert(r.left, x)
    else:
        r.right = insert(r.right, x)

    return r 
```

```python
root = None 

for x in [50, 30, 70, 20, 40, 60, 80]:
    root = insert(root, x)
```

```python
def contains(r, x):
    if not r:
        return False 

    if r.x == x:
        return True 

    if x < r.x:
        return contains(r.left, x)

    return contains(r.right, x)

print(contains(root, 40))
print(contains(root, 90))
```

```python
def inorder(r):
    if r:
        inorder(r.left)
        print(r.x, end=' ')
        inorder(r.right)

inorder(root)
```

```python
def preorder(r):
    if r:
        print(r.x, end=' ')
        preorder(r.left)
        preorder(r.right)

preorder(root)
```

```python
def postorder(r):
    if r:
        postorder(r.left)
        postorder(r.right)
        print(r.x, end=' ')

postorder(root)
```

# DFS Graph Traversal

```python
g = {
    0: [1, 2],
    1: [3],
    2: [4],
    3: [],
    4: []
}
```

```python
from collections import deque

def bfs(g, start):
    q = deque([start])
    seen = {start}

    while q:
        u = q.popleft()
        print(u, end=" ")

        for v in g[u]:
            if v not in seen:
                seen.add(v)
                q.append(v)

bfs(g, 0)
```

# BFS Graph Traversal

```python
def dfs(g, u, seen=set()):
    if u in seen:
        return 

    seen.add(u)
    print(u, end=' ')

    for v in g[u]:
        dfs(g, v, seen)

dfs(g, 0)
```
