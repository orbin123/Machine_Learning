### Q3. Simple Linear Regression using Gradient Descent

```python
import numpy as np 

X = np.array([1, 2, 3, 4,5])
y = np.array([1, 4, 6, 8, 10])

w = 0
b = 0
lr = 0.01

for i in range(1000):
    y_pred = w * X + b 

    dw = (-2/len(X)) * np.sum(X * (y-y_pred))
    db = (-2/len(X)) * np.sum(y-y_pred)

    w -= lr * dw
    b -= lr * db 

print("w =", w)
print("b =", b)
```

### Q5. Visulaize Gradient Descent Steps

```python
import numpy as np 
import matplotlib.pyplot as plt 

X = np.array([1, 2, 3])
y = np.array([2, 4, 6])

w = 0
b = 0
lr = 0.1 

weights = []
costs = []

for i in range(20):
    y_pred = w * X + b 

    cost = np.mean((y-y_pred) ** 2)

    dw = (-2/len(X)) * np.sum(X * (y-y_pred))
    db = (-2/len(X)) * np.sum(y-y_pred)

    weights.append(w)
    costs.append(cost)

    w -= lr * dw
    b -= lr * db 

plt.plot(weights, costs)
plt.xlabel("w")
plt.ylabel("Cost")
plt.show()
```

### Z-Test from scratch

**Formula:** `z = (x̄ - μ) / (σ / √n)`

```python
import math 
from scipy.stats import norm 

x, mu, sigma, n = 105, 100, 15, 36

z = (x-mu) / (sigma / math.sqrt(n))
p = 2*(1-norm.cdf(abs(z)))

print(z, p)
```

```python
if p < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

### T-Test from scratch

**Formula:** `t = (x̄ - μ) / (s / √n)`

```python
data = [12, 15, 14, 10, 13, 15, 11, 14]

import numpy as np 
from scipy.stats import t 

data = np.array(data)
mu = 12 

tval = (data.mean() - mu) / (data.std(ddof=1) / np.sqrt(len(data)))
p = 2 * (1-t.cdf(abs(tval), len(data) - 1))

print(tval, p)
```

### Chi-Square Goodness-of-Fit Test

Observed frequencies are `[20, 30, 50]`.

Expected frequencies are `[25, 25, 50]`.

```python
from scipy.stats import chisquare

obs = [20, 30, 50]
expected = [25, 25, 50]

chi, p = chisquare(obs, expected)

print(chi, p)
```

```python
if p < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

### Chisquare from scratch

```python
import numpy as np 
from scipy.stats import chi2 

def chi_square(observed):
    obs = np.array(observed, dtype=float)

    # Sums 
    row_totals = obs.sum(axis=1)
    col_totals = obs.sum(axis=0)
    grand_total = obs.sum()

    expected = np.outer(row_totals, col_totals)/grand_total

    chi_square_stat = np.sum(((obs - expected)**2) /expected)

    dof = (obs.shape[0] -1) * (obs.shape[1] - 1)

    p_value = chi2.sf(chi_square_stat, dof)

    return chi_square_stat, p_value, dof, expected

observed_data = [
    [50, 10], 
    [30, 40]
]

stat, p, df, expected = chi_square(observed_data)

print(stat, p,df, expected)
```

### ANOVA from Scratch

```python
A = np.array([10, 12, 11, 13])
B = np.array([15, 16, 14, 17])
C = np.array([20, 19, 21, 22])
```

```python
from scipy import stats 

groups = [A, B, C]
grand = np.mean(np.concatenate(groups))

K = len(groups)
N = sum(len(group) for group in groups)

df_between, df_within = K-1, N-K 

SSB = sum(len(g) * (g.mean() - grand) ** 2 for g in groups)
SSW = sum(sum((g-g.mean())**2) for g in groups)

F = (SSB/df_between)/(SSW/df_within)

print(F)

p = stats.f.sf(F, df_between, df_within)

print(f"p-value: {p:.7f}")

if p < 0.05:
    print("Reject H0")
else:
    print("Fail to reject H0")
```
