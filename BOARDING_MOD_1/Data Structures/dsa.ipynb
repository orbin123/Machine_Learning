```python
def rotate(nums: list[int], k: int) -> None:
    """Rotate nums right by k steps in place (modifies nums)."""

    n = len(nums)
    k %= n
    if k == 0:
        return 

    def reverse(lo: int, hi: int) -> None:
        while lo < hi:
            nums[lo], nums[hi] = nums[hi], nums[lo]
```
