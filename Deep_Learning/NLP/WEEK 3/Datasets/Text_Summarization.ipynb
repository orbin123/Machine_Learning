# Extractive Summarization using TextRank

```python
# TextRank implementation

import nltk
nltk.download("punkt")
nltk.download("punkt_tab")

text = """
Machine learning is a field of artificial intelligence that focuses on teaching machines how to learn from data.
It is widely used in applications such as recommendation systems, image recognition, and speech processing.
Text summarization is an important NLP task that helps reduce reading time.
Extractive summarization selects important sentences directly from the text.
"""
```

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer


parser = PlaintextParser.from_string(text, Tokenizer("english"))
summarizer = TextRankSummarizer()

summary = summarizer(parser.document, sentences_count=2)
```

```python
for sentence in summary:
    print(sentence)
```

# Abstractive Summarization using BERT

```python
from transformers import pipeline

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)


```

```python
text = """
Machine learning is a type of artificial intelligence that focuses on building a lot of systems that learn from data.
These systems can be improved automatically through experience and multiple training.
Text summarization helps compress or summarize long documents into small versions while keeping key information on it.
Abstractive summarization generates new sentences rather than just simply copying from the original text or data.
"""

```

```python
summary = summarizer(
    text,
    max_length=60,
    min_length=25,
    do_sample=False
)

print("BERT Summary:\n")
print(summary[0]["summary_text"])
```
