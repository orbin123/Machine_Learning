```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import stopwords, wordnet
from nltk.text import TextCollection
from nltk import pos_tag, ne_chunk
import string

# ============================================================
# 0. DOWNLOADS & SETUP
# ============================================================
# NLTK requires specific data packages for its algorithms
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)
nltk.download('wordnet', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('maxent_ne_chunker', quiet=True)
nltk.download('words', quiet=True)

# Dataset
corpus = [
    "I'm learning NLP! It's amazing.",
    "Natural Language Processing is fascinating.",
    "Python is great for NLP tasks."
]

print("DATASET:", corpus)

# ============================================================
# 1. BASIC PREPROCESSING (Standard NLTK)
# ============================================================

# 1a. TOKENIZATION
text = corpus[0]
word_tokens = word_tokenize(text)
sent_tokens = sent_tokenize(text)
print("\n--- Basic NLTK ---")
print("Word Tokens:", word_tokens)

# 1b. STEMMING
porter = PorterStemmer()
snowball = SnowballStemmer('english')
words = ['running', 'runs', 'easily']
print("Porter Stem:", [porter.stem(w) for w in words])
print("Snowball Stem:", [snowball.stem(w) for w in words])

# 1c. LEMMATIZATION
lemmatizer = WordNetLemmatizer()
# Note: WordNetLemmatizer works best if you provide the POS tag (default is Noun)
print("Lemmatized:", [lemmatizer.lemmatize(w, pos='v') for w in words])

# ============================================================
# 2. ADVANCED CLEANING (Using NLTK logic)
# ============================================================

contractions = {
    "I'm": "I am", "It's": "It is", "don't": "do not",
    "can't": "cannot", "won't": "will not"
}

def nltk_preprocess(text):
    # Expand contractions
    for cont, exp in contractions.items():
        text = text.replace(cont, exp)
    
    text = text.lower()
    tokens = word_tokenize(text)
    
    stop_words = set(stopwords.words('english'))
    # Filter punctuation and stopwords
    clean_tokens = [
        lemmatizer.lemmatize(w) 
        for w in tokens 
        if w not in string.punctuation and w not in stop_words
    ]
    
    return clean_tokens

# Clean the whole corpus
cleaned_corpus_tokens = [nltk_preprocess(doc) for doc in corpus]
print("\n--- Cleaned Corpus ---")
print(cleaned_corpus_tokens)

# ============================================================
# 3. POS TAGGING & NER (Replacing SpaCy)
# ============================================================

print("\n--- NLTK POS Tagging & NER ---")
for sent_tokens in cleaned_corpus_tokens:
    # POS Tagging
    tagged = pos_tag(sent_tokens)
    
    # NER (Named Entity Recognition)
    # ne_chunk creates a tree; we print it simply here
    entities = ne_chunk(tagged)
    
    print(f"Tags for {sent_tokens}: {tagged}")
    # Note: 'binary=True' in ne_chunk would just show "NE" instead of "PERSON/ORG"

# ============================================================
# 4. TF-IDF VECTORIZATION (Replacing Sklearn)
# ============================================================
# NLTK uses TextCollection for calculating TF-IDF stats
print("\n--- NLTK TF-IDF ---")

# Flatten list of lists for the collection context
tc = TextCollection(cleaned_corpus_tokens)

target_word = 'nlp'
print(f"Stats for word '{target_word}':")

# Calculate IDF (Inverse Document Frequency) - Global importance
print(f"IDF score: {tc.idf(target_word):.4f}")

# Calculate TF-IDF for the word 'nlp' inside the first document
# Note: The first doc is cleaned_corpus_tokens[0] -> ['learning', 'nlp', 'amazing']
tf_idf_score = tc.tf_idf(target_word, cleaned_corpus_tokens[0])
print(f"TF-IDF in Doc 1: {tf_idf_score:.4f}")

# ============================================================
# 5. SEMANTIC SIMILARITY (Replacing Word2Vec/SpaCy)
# ============================================================
# Since NLTK lacks native neural Word2Vec, we use WordNet (Wu-Palmer Similarity)
# to measure semantic distance between concepts.

print("\n--- NLTK Semantic Similarity (WordNet) ---")

w1 = wordnet.synsets('python')[0]   # Get first definition of Python (snake/language)
w2 = wordnet.synsets('computer')[0] # Get first definition of Computer
w3 = wordnet.synsets('snake')[0]

# Wu-Palmer Similarity returns a score 0-1 based on hierarchy in the dictionary
sim_python_computer = w1.wup_similarity(w2)
sim_python_snake = w1.wup_similarity(w3)

print(f"Similarity (Python vs Computer): {sim_python_computer:.4f}")
print(f"Similarity (Python vs Snake): {sim_python_snake:.4f}")
```
