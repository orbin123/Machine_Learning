### Practical Task: Text Preprocessing

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import stopwords 
import re 
import pandas as pd

stop_words = stopwords.words('english')
```

```python
data = pd.read_csv('tripadvisor_hotel_reviews.csv')
data.info()
```

```python
data.head()
```

```python
data['review_lowercase'] = data['Review'].str.lower()
```

```python
data['review_lowercase'].head()
```

```python
stop_words.remove('not')

data['review_stpwds_rm'] = data['review_lowercase'].apply(lambda x: ' '.join([word for word in x.split() if word not in (stop_words)]))

data['review_stpwds_rm'][2]
```

```python
data['review_no_punct'] = data.apply(lambda x: re.sub(r'[*]', 'star', x['review_stpwds_rm']), axis=1)
data['review_no_punct'][2]
```

```python
data['review_no_punct'] = data.apply(lambda x: re.sub(r'([^\w\s])', '', x['review_stpwds_rm']), axis=1)
data['review_no_punct'][2]
```

```python
data['tokenized'] = data.apply(lambda x: word_tokenize(x['review_no_punct']), axis=1)
data['tokenized'][2]
```

```python
data['stemmed'] = data['tokenized'].apply(lambda tokens: [ps.stem(token) for token in tokens])

data.head()
```

```python
wnl = WordNetLemmatizer()
data['lemmatized'] = data['stemmed'].apply(lambda tokens: [wnl.lemmatize(token) for token in tokens])
data[['tokenized', 'stemmed', 'lemmatized']].head()
```

```python
tokens_clean = sum(data['lemmatized'], [])
```

```python
unigrams = (pd.Series(nltk.ngrams(tokens_clean, 1)).value_counts())
print(unigrams)
```

```python
bigrams = (pd.Series(nltk.ngrams(tokens_clean, 2)).value_counts())
print(bigrams)
```

### Practical Task: POS and NER

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords 
from nltk.stem import WordNetLemmatizer 
import spacy 
import re 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
bbc_data = pd.read_csv('bbc_news.csv')
```

```python
bbc_data.head()
```

```python
bbc_data.info()
```

```python
titles = pd.DataFrame(bbc_data['title'])
titles.head()
```

```python
# clean data 
# lowercase 
titles['lower_case'] = titles['title'].str.lower()
titles.head()
```

```python
# stopwords removal 
stop_words = stopwords.words('english')
titles['no_stopwords'] = titles['lower_case'].apply(lambda x: ' '.join([word for word in x.split() if word not in (stop_words)]))
titles.head()
```

```python
# Punctuation Removal 
punt = r"([^\w\s])"
titles["no_punct"] = titles.apply(lambda x: re.sub(punt, '', x['no_stopwords']), axis=1)
titles.head()
```

```python
# tokenize 
titles['raw_tokenized'] = titles.apply(lambda x: word_tokenize(x['title']), axis=1)
titles['tokenized'] = titles.apply(lambda x: word_tokenize(x['no_punct']), axis=1)
titles.head()
```

```python
wnl = WordNetLemmatizer()
titles['lemmatized'] = titles['tokenized'].apply(lambda tokens: [wnl.lemmatize(token) for token in tokens ])
titles.head()
```

```python
tokens_raw_list = sum(titles['raw_tokenized'], [])
tokens_clean_list = sum(titles['lemmatized'], [])
```

#### POS Tagging

```python
nlp = spacy.load('en_core_web_sm')

spacy_doc = nlp(' '.join(tokens_raw_list))

pos_df = pd.DataFrame(columns=['token', 'pos_tag'])

for token in spacy_doc:
    pos_df = pd.concat([pos_df, pd.DataFrame.from_records([{'token': token.text, 'pos_tag': token.pos_}])], ignore_index=True)
```

```python
pos_df_counts = pos_df.groupby(['token', 'pos_tag']).size().reset_index(name='counts').sort_values(by='counts',ascending=False)
```

```python
pos_df_counts.head(15)
```

```python
nouns = pos_df_counts[pos_df_counts.pos_tag =='NOUN'][0:10]
nouns
```

#### NER Tagging

```python
ner_df = pd.DataFrame(columns=['token', 'ner_tag'])

for token in spacy_doc.ents:
    if pd.isna(token.label_) is False:
        ner_df = pd.concat([ner_df, pd.DataFrame.from_records(
            [{'token': token.text, 'ner_tag': token.label_}]
        )], ignore_index=True)
```

```python
ner_df_counts = ner_df.groupby(['token', 'ner_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
```

```python
ner_df_counts.head()
```

### Vectorization

#### Bag of Words

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

data = [' Most shark attacks occur about 10 feet from the beach since that is where the people are',
        'the efficiency with which he paired the socks in the drawer was quite admirable',
        'carol drank the blood as if she were a vampire',
        'giving directions that the mountains are to the west only works when you can see them',
        'the sign said there was road work ahead so he decided to speed up',
        'the gruff old man sat in the back of the bait shop grumbling to himself as he scooped out a handful of worms']
```

```python
countvec = CountVectorizer()

countvec_fit = countvec.fit_transform(data)

bag_of_words = pd.DataFrame(countvec_fit.toarray(), columns=countvec.get_feature_names_out())

bag_of_words.head()
```

#### TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf_idf_vec = TfidfVectorizer()

tf_idf_fit = tf_idf_vec.fit_transform(data)

tfidf_bag = pd.DataFrame(tf_idf_fit.toarray(), columns=tf_idf_vec.get_feature_names_out())

tfidf_bag.head()
```
