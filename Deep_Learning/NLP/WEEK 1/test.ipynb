```python
import nltk
import string
import contractions
import pandas as pd
from textblob import TextBlob
from nltk.tokenize import sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import stopwords 
from sklearn.feature_extraction.text import TfidfVectorizer


df = pd.read_csv('nlp_sample_dataset.csv')
```

```python
df.head()
```

```python
df['lowercase'] = df['text'].str.lower()
df.head()
```

```python
df['punctuation'] = df['lowercase'].apply(lambda x: ''.join([c for c in x if c not in string.punctuation]))
df.head()
```

```python
df['mispelling'] = df.apply(lambda x: str(TextBlob(x['punctuation']).correct()))
df.head()
```

```python
df['contractions'] = df['mispelling'].apply(contractions.fix)
df.head()
```

```python
df['stopwords'] = df['contractions'].apply(lambda x: [w for w in x if w not in stopwords.words('english')])
df.head()
```

```python
ps = PorterStemmer()
sbs = SnowballStemmer()
wnl = WordNetLemmatizer()

for w in df['lowercase'][0]:
    print(ps.stem(w), sbs.stem(w), wnl.lemmatize(w))
```

```python
import spacy 

nlp = spacy.load('en_core_web_sm')

doc = nlp(df['lowercase'][0])

for token in doc:
    print(token.text, token.pos_, token.lemma_, token.ent_type_)
    
```

```python
sentence = [w for w in df['lowercase'][0]]
tf_idf = TfidfVectorizer()
tf_idf_fit = tf_idf.fit_transform([sentence])

from sklearn.metrics.pairwise import cosine_similarity

print(cosine_similarity(tf_idf_fit[0], tf_idf_fit[1]))
```
